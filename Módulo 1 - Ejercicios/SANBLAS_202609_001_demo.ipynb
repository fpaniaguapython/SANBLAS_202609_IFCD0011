## Importaciones de librerías

```python
import pandas as pd
```

## Obtención del **fichero de datos**

```python
URL_DATA_FILE = 'https://fpaniaguapython.github.io/datos/sanblas202609.csv'
df_original = pd.read_csv(URL_DATA_FILE)
df_original
```

Creo una copia del df_original para trabajar con ella y tener siempre disponible el df_original

```python
df_trabajo = df_original.copy()
```

Sustituir de los nombres de columnas por otros más sencillos

```python
df_trabajo.columns = df_trabajo.columns.str.replace(' ', '_').str.lower()
df_trabajo.head() # Muestra los (por defecto) 5 primeros registros
```

## Anonimizar la serie

```python
df_trabajo['nombre_y_apellidos']='Anónimo'
df_trabajo.head()
```
